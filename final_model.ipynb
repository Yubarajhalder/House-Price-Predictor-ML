```python
import joblib
import pandas as pd
```

```python
model=joblib.load("final_model.pkl")
```

```python
# Inputs
bd = int(input("Enter Number Of Bedrooms: "))

print("0.5 -> toilet + sink")
print("1.0 -> toilet + sink + shower/bath")
bathroom = float(input("Enter Bathrooms: "))

sq_ft = int(input("Enter Square Feet Living Area: "))
sq_lot = int(input("Enter Total Lot Area: "))

if sq_ft <= sq_lot:

    floor = float(input("Enter Floors: "))

    condition = int(input("Enter Condition (1-5): "))

    view = int(input("Enter View Rating: "))

    sqft_above = int(input("Enter sqft_above: "))

    sqft_basement = int(input("Enter sqft_basement: "))

    yr_built = int(input("Enter Year Built: "))

    print("\nYear Categories")
    print("1 -> Very Old House")
    print("2 -> Old House")
    print("3 -> Medium Age")
    print("4 -> New House")
    print("5 -> Latest House")

    year_cat = int(input("Enter Year Category: "))

    city = input("Enter City: ")

    # DataFrame for prediction
    x = pd.DataFrame([{
        "bedrooms": bd,
        "bathrooms": bathroom,
        "sqft_living": sq_ft,
        "sqft_lot": sq_lot,
        "floors": floor,
        "view": view,
        "condition": condition,
        "sqft_above": sqft_above,
        "sqft_basement": sqft_basement,
        "yr_built": yr_built,
        "year_cat": year_cat,
        "city": city
    }])

    # Prediction
    output = model.predict(x)

    print("\nPredicted House Price = $", round(output[0], 2))

else:
    print("Invalid: sqft_living cannot be greater than sqft_lot")
```
